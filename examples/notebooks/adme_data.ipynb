```python
import pandas as pd
from rdkit.Chem import MolFromSmiles, Descriptors
```

```python
df = pd.read_csv("ADME_public_set_3521.csv")
df
```

```python
def featurise(smiles: str) -> dict:
    mol = MolFromSmiles(smiles)
    return pd.Series(Descriptors.CalcMolDescriptors(mol))
```

```python
sol_df = df[["SMILES", "LOG SOLUBILITY PH 6.8 (ug/mL)"]].dropna()
sol_df
```

```python
sol_features = sol_df["SMILES"].apply(featurise)
```

Strip descriptors with any 0 values
